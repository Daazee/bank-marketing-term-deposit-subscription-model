```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
import joblib
```

```python
df = pd.read_csv('../data/interim/bank-interim.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.replace(".", "_").str.lower()
```

```python
df.info()
```

```python
num_cols = df.select_dtypes(include='number').columns.tolist()

cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
print(num_cols)
print(cat_cols)
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42, stratify=df.y)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42, stratify=df_full_train.y)  # 0.25 x 0.8 = 0.2
```

```python
df_train = df_train.fillna(0)
```

```python
df_train.info()
```

```python
train_dicts = df_train.drop(columns="y").to_dict(orient='records')
y_train = df_train.y
```

```python
val_dicts = df_val.drop(columns="y").to_dict(orient='records')
y_train = df_val.y
```

```python
test_dicts = df_test.drop(columns="y").to_dict(orient='records')
y_test = df_test.y
```

```python
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
```

```python
# No fiiting required as fitting is already done on training data
X_val = dv.transform(val_dicts)
X_test = dv.transform(test_dicts)
```

```python
# Convert to a new DataFrame to view the encoded features
processed_training_df = pd.DataFrame(X_train, columns=dv.get_feature_names_out())
processed_training_df.head()
```

```python
# Convert to a new DataFrame to view the encoded features
processed_val_df = pd.DataFrame(X_val, columns=dv.get_feature_names_out())
processed_val_df.head()
```

```python
# Convert to a new DataFrame to view the encoded features
processed_test_df = pd.DataFrame(X_test, columns=dv.get_feature_names_out())
processed_test_df.head()
```

```python
# check whether y is already in encoded_df  dataframe
# no means we did not include y in our DictVectorizer. That's what we want
['yes' if 'y' in processed_training_df.columns.tolist() else 'no']
```

```python
print("Processed Training DataFrame Shape:", len(processed_training_df))
print("Original Training DataFrame Shape:", len(df_train))

print("Processed Validation DataFrame Shape:", len(processed_val_df))
print("Original Validation DataFrame Shape:", len(df_val))

print("Processed Test DataFrame Shape:", len(processed_test_df))
print("Original Test DataFrame Shape:", len(df_test))
```

```python
# Without to_numpy method, it will not able to match the index since encoded was created from numpy arrays. The y column is a pandas Series. We want it to be a numpy array for model training.
processed_training_df["y"] = (df_train["y"] == "yes").astype(int).to_numpy()
```

```python
processed_val_df["y"] = (df_val["y"] == "yes").astype(int).to_numpy()
processed_test_df["y"] = (df_test["y"] == "yes").astype(int).to_numpy()
```
