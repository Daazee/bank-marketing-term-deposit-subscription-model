```python
global_subscribe_rate = df.y.mean()
global_subscribe_rate
```

```python
from IPython.display import display

for c in cat_cols:
    print(c)
    df_group = df.groupby(c).y.agg(['mean', 'count'])
    df_group['diff'] = df_group['mean'] - global_subscribe_rate
    df_group['risk'] = df_group['mean'] / global_subscribe_rate
    display(df_group)
    print()
    print()
```
