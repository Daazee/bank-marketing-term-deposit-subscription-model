```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
import xgboost as xgb
import joblib
from datetime import datetime
```

```python
df_train = pd.read_csv('../data/processed/bank-processed.csv')
```

```python
df_train.head()
```

```python
X_train = df_train.drop('y', axis=1).to_numpy()
y_train = df_train['y'].to_numpy()
```

```python
X_train
```

```python
# Using logistic regression as a baseline model
logistic_model = LogisticRegression(max_iter=1000)
logistic_model.fit(X_train, y_train)
```

```python
joblib.dump(logistic_model, f'../artifacts/logistic_model_{datetime.now().strftime("%Y-%m-%d_%H-%M-%S")}.bin')
```
